## 4. 微积分



### 4.1 导数



```python
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

def f(x):
    return x ** 2 + 3 * x + 2

x = 2.0
print(f"数值导数 f'({x}) = numerical_derivative(f, x):.4f")
```

### 4.2 偏导数

多元函数中，只对一个变量求导，其他的变量固定

```python
import numpy as np

def partial_derivative(f, x, var_idx, h=1e-5):
    """f 是多元函数，var_idx 是要求导的变量索引"""
    x_plus = x.copy()
    x_minus = x.copy()
    x_plus[var_idx] += h
    x_minus[var_idx] -= h
    return (f(x_plus) - f(x_minus)) / (2 * h)

def f(x):
    return x[0]**2 + x[0]*x[1] + x[1]**2

x = np.array([1.0, 2.0])
print(partial_derivative(f, x, 0))
print(partial_derivative(f, x, 1))
```

### 4.3 梯度



```python
# 1. 定义需要求梯度的变量
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)

# 2. 定义函数 y = f(x)
y = x[0]**2 + x[1]**2 + x[2]**2

# 3. 反向传播，自动计算梯度
y.backward()

# 4. 查看梯度
print(f"∇_x y = {x.grad}")
# 解析验证：∂y/∂x_1 = 2x_1 = 2, ∂y/∂x_2 = 2x_2 = 4, ∂y/∂x_3 = 2x_3 = 6
```

矩阵梯度

```python
# y = ||Xw - y_true||^2  (最小二乘损失)
X = torch.tensor([[1., 2.],
                  [3., 4.],
                  [5., 6.]])
w = torch.tensor([0.1, 0.2], requires_grad=True)
y_true = torch.tensor([1., 2., 3.])

y = torch.norm(X @ w - y_true) ** 2
y.backward()

print(f"∇_w y = {w.grad}")

# ∇_w ||Xw-y||^2 = 2X^T(Xw-y)
```

### 定积分

```python
import sympy as sp
#法1
from scipy import integrate

def f(x):
    return x ** 2  # f(x) = x^2

result, error = integrate.quad(f, 0, 1)
print(f"∫_0^1 x^2 dx = {result:.6f} (误差估计: {error:.2e})")


#法2
x = sp.Symbol('x')
f_sym = x ** 2
integral = sp.integrate(f_sym, (x, 0, 1))
print(f"符号积分结果: {integral}")
```

```python
import torch
import numpy as np
from scipy import stats, integrate
import sympy as sp
# 定义分布
normal_dist = stats.norm(loc=0, scale=1)  # N(0, 1)

# 采样 z ~ P
z = normal_dist.rvs(size=5)  # 采集5个样本
print(f"z ~ N(0,1): {z}")

# PyTorch 采样
z_torch = torch.randn(5)  # 标准正态
print(f"z ~ N(0,1) (torch): {z_torch}")
```

### 5.2 条件概率



```python
import torch
import numpy as np

# 离散条件概率示例
# 假设：P(下雨|阴天) = P(下雨且阴天) / P(阴天)

# 联合概率表（模拟）
#        下雨   不下雨
# 阴天    0.15   0.25
# 晴天    0.05   0.55

joint = np.array([[0.15, 0.25],
                  [0.05, 0.55]])

# P(阴天) = 0.15 + 0.25 = 0.40
P_cloudy = joint[0, :].sum()

# P(下雨|阴天) = 0.15 / 0.40 = 0.375
P_rain_given_cloudy = joint[0, 0] / P_cloudy
print(f"P(下雨|阴天) = {P_rain_given_cloudy:.3f}")
```

### 5.4 数学期望



```python
# 离散期望：掷骰子 E[X]
outcomes = np.array([1, 2, 3, 4, 5, 6])
probs = np.ones(6) / 6  # 均匀分布
E_X = (outcomes * probs).sum()
print(f"E[X] = {E_X:.2f}")  # 3.50

# 连续期望：E[X] where X ~ N(0,1)，用蒙特卡洛近似
samples = np.random.normal(0, 1, size=100000)
E_X_mc = samples.mean()
print(f"E[X] (Monte Carlo) ≈ {E_X_mc:.4f}")  # 接近 0

# 函数期望：E[X^2] where X ~ N(0,1)
E_X2 = (samples ** 2).mean()
print(f"E[X^2] ≈ {E_X2:.4f}")  # 接近 1（方差）
```

### 5.5 独立性

```python
# 检验独立性：计算协方差或相关系数
X = np.random.randn(1000)
Y = np.random.randn(1000)  # Y 独立生成

# 独立变量的协方差应接近0
cov = np.cov(X, Y)[0, 1]
print(f"Cov(X,Y) = {cov:.4f}")

# 相关系数
corr = np.corrcoef(X, Y)[0, 1]
print(f"ρ(X,Y) = {corr:.4f}")
```

### 5.6 条件独立

```python
# 条件独立示例：
# X = 考试成绩, Y = 冰淇淋销量, Z = 温度
# 给定温度 Z，考试成绩和冰淇淋销量条件独立；把 Z "按住不动"，X 和 Y 就"各走各的路"，不再相关。

# 模拟数据
np.random.seed(42)
Z = np.random.normal(25, 5, 1000)  # 温度
X = 0.5 * Z + np.random.randn(1000)  # 成绩（受温度影响）
Y = 2.0 * Z + np.random.randn(1000)  # 冰淇淋销量（受温度影响）

# 无条件时 X 和 Y 相关
print(f"无条件 ρ(X,Y) = {np.corrcoef(X, Y)[0,1]:.4f}")

# 条件独立检验（简化：固定 Z 的某个范围）
mask = (Z > 24) & (Z < 26)  # 固定温度在 24-26
print(f"条件 ρ(X,Y|Z∈[24,26]) = {np.corrcoef(X[mask], Y[mask])[0,1]:.4f}")
# 做数据分析时不控制温度，就会得出"成绩和冰淇淋相关"这种荒谬结论。
```

### 5.7 方差与标准差

### 5.3 协方差与相关系数



```python
# 生成相关数据
X = np.random.randn(1000)
Y = 2 * X + np.random.randn(1000)  # Y 与 X 正相关

cov = np.cov(X, Y)[0, 1]
print(f"Cov(X,Y) = {cov:.4f}")

corr = np.corrcoef(X, Y)[0, 1]
print(f"ρ(X,Y) = {corr:.4f}")  # 接近 0.89

# 手动验证
mean_x, mean_y = X.mean(), Y.mean()
cov_manual = ((X - mean_x) * (Y - mean_y)).mean()
corr_manual = cov_manual / (X.std() * Y.std())
print(f"手动 ρ = {corr_manual:.4f}")
```

### 5.9 熵



```python
# 离散熵
# 公平硬币：p = [0.5, 0.5]
# H = -0.5*log(0.5) - 0.5*log(0.5) = log(2) ≈ 0.693
p_fair = np.array([0.5, 0.5])
H_fair = stats.entropy(p_fair, base=np.e)  # base = np.e表示使用自然对数
print(f"H(公平硬币) = {H_fair:.4f}")  # 0.6931 = ln(2)

# 有偏硬币：p = [0.9, 0.1]
p_biased = np.array([0.9, 0.1])
H_biased = stats.entropy(p_biased)
print(f"H(有偏硬币) = {H_biased:.4f}")  # 0.3215更小，因为确定性更高

# 确定性分布：p = [1, 0]
p_det = np.array([1.0, 0.0])
H_det = stats.entropy(p_det)
print(f"H(确定) = {H_det:.4f}")  # 0
```

### 5.10 KL散度



```python
P = np.array([0.4, 0.3, 0.2, 0.1])
Q = np.array([0.25, 0.25, 0.25, 0.25])  # 均匀分布

# KL(P || Q)
kl_pq = stats.entropy(P, Q)
print(f"D_KL(P||Q) = {kl_pq:.4f}")

# KL(Q || P) — 不对称！
kl_qp = stats.entropy(Q, P)
print(f"D_KL(Q||P) = {kl_qp:.4f}")

# 手动验证
kl_manual = np.sum(P * np.log(P / Q))
print(f"手动 D_KL = {kl_manual:.4f}")

# 当 P = Q 时，KL = 0
print(f"D_KL(P||P) = {stats.entropy(P, P):.4f}")  # 0

P = np.array([0.8, 0.2])
Q1 = np.array([0.75, 0.25])
Q2 = np.array([0.5, 0.5])
Q3 = np.array([0.2, 0.8])  # 完全反了

print(stats.entropy(P, Q1))  # 0.0070  ← 很小，预测很准
print(stats.entropy(P, Q2))  # 0.1927  ← 中等，预测一般
print(stats.entropy(P, Q3))  # 0.8318  ← 很大，预测离谱
```

KL 散度 D_KL(P || Q)：

- Q 和 P 差了多少；
- 或者说，用 Q 代替 P，多损失了多少信息。

D_KL(P || Q) = ∑ P(i) log(P(i) / Q(i))

##### 性质
 D_KL(P || Q) ≥ 0

D_KL(P || Q) = 0 当且仅当 P = Q

D_KL(P || Q) ≠ D_KL(Q || P)
