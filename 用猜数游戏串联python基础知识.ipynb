合并为一个函数

```python
import random

def play_game():
    while True:
        secret = random.randint(1,10)
        print("I have chosen number from 1 to 10, can you try?")

        attempts = 0

        while True:
            guess = int(input("please input your guess:"))
            attempts += 1

            if guess > secret:
                print("too big")
            elif guess < secret:
                print("too small")
            else:
                print("you win!you have tried for{attempts}times")
                break

        play_again = input("do you want to play again?(y/n)").strip().lower()
        if play_again != "y":
            print("thanks for playing")
            break

play_game()
```

拆分函数

code implementation

```python
import random

def get_user_guess():
    return int(input("please input your guess:"))

def check_guess(guess, secret):
    if guess > secret:
        print("too big")
    elif guess < secret:
        print("too small")
    else:
        print("equal")

def play_game():
    secret = random.randint(1, 10)
    print("I have chosen number from 1 to 10, can you try?")
    guess = None
    while guess != secret:
        guess = get_user_guess()
        result = check_guess(guess, secret)
        if result == "too big":
            print("too big")
        elif result == "too small":
            print("too small")
        else:
            print("bingo")

play_game()
print("Game over")
```

## 异常处理

```python
import random

while True:
    secret_number = random.randint(1, 10)
    print("I have chosen number from 1 to 10, can you try?")

    attempts = 0

    while True:
        try:
            guess = int(input("please input your guess:"))
            attempts += 1
            if guess > secret_number:
                print("too big")
            elif guess < secret_number:
                print("too small")
            else:
                print("you win!you have tried for{attempts}times")
                break
        except ValueError:
            print("please input a vaild number")

    play_again = input("do you want to play again?(y/n)").strip().lower()
    if play_again != "y":
        print("thanks for playing")
        break
```
