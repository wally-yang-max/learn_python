## 文件读写的两种模型
1. 文件不存在：1新建文件，2写入信息，3关闭保存文件
1. 文件存在：1打开文件，2读取信息， 3关闭文件
1. open() 函数：常用形式是接收两个参数文件名(file)和模式(mode)
-  r ，只读模式【默认模式，文件必须存在，不存在则抛出异常】
-  w，只写模式【不可读；不存在则创建；存在则清空内容】
-  a，只读追加写模式【不可读；不存在则创建；存在则只追加内容】
-  r+ 表示可以同时读写某个文件
-  r+， 读写【可读，可写】
-  w+，写读【可读，可写】
-  a+， 写读【可读，可写】


```python
#创建文件 此时为文件不存在的情况
data_file = open('love_study.txt', 'w')
str1 = "love makes the world go round!  "
str2 = "good good study day day up! "
# 2 写入文件
line = data_file.write(str1)
line = data_file.write(str2) #写入文件
# 3 关闭保存文件
data_file.close()

# 文件存在 1打开文件，2读取信息， 3关闭文件
# 1 打开文件
fo = open("love_study.txt", "r") #读取文件
# 2 读取文件
line = fo.read()
print ("读取的字符串: %s" % (line))
# 3 关闭文件
fo.close()
```

## with语句
1. with语句使代码更清晰、更具可读性， 它简化了文件读写的流程；
1. with关键字系统会自动调用 f.close() 方法

```python
import sympy as sp

#解方程 x^2 - 8x + 12 = 0
x = sp.symols('x')
roots = sp.solve(x**2 - 8*x + 12, x)
print("Roots:", roots)

with open("quadratic_roots.txt", "w") as file:
    if isinstance(roots, tuple):
        file.write(f"Root 1: {roots[0]}\nRoot 2: {roots[1]}")
    else:
        file.write(f"Root: {roots}")

fo = open("quadratic_roots.txt", "r")
line = fo.read()
print("读取的字符串：%s" % line)
fo.close()
```

### 三种错误
- 语法错误：比如少写了一个引号，或写中文的引号。编写了不符合Python语法的代码所引起的错误。
- 运行时错误：比如用一个数去除以零。代码运行期间发生的错误称为异常或运行时错误。
- 逻辑错误：比如解一元二次方程的根x1 = -b + sqrt_delta /2*a 忘记写括号 。逻辑错误发生时程序会顺利运行，但得到的结果不符合期望。
- 常见的错误
* IndentationError：缩进错误
* SyntaxError：语法错误
* invalid character in identifier：可能输入中文符号。

```python
######  除法函数的异常处理操作
# try:
#      # 可能出现的异常代码
#except 异常类型1:
       #处理异常类型1
# except 异常类型2:
        # 处理异常类型2:
# else:
#       # 没有异常时执行
# finally:
#       # 无论是否出现异常都会执行
def divide(x, y):
    '''除法'''
    try:
        print("1可能发生异常的代码")
        result = x / y
        print(result)
    except ZeroDivisionError as err:#除0的错误
        print('2发生异常时执行的代码，未知错误', err)
    except Exception as e:#大类异常错误
        print("2发生异常时执行的代码，异常错误", e)
    else:
        print("3没有发生异常执行的代码！")
    finally:
        print("4非必须代码，不管是是否有异常都会执行！")
    print("我是正常代码！")

divide(2, 1)
print("---------------------")
divide(2, 0)
print("-----------------------")
divide(2, 'a')
```

求一元二次方程的根，添加异常处理

```python
import math
def solve_equation(a, b, c):
    try:
        if a == 0:
                print(f'您输入的不是二次方程!!!一次方程的解为x={c/b}')
        else:
            # 计算判别式（discriminant）
            delta = b*b-4*a*c
            if delta == 0:
                # 有一个实根（重根）
                x = -b/(2*a)
                print(f'方程有两个相等的实数根:x={x}')
                return x
            elif delta > 0:
                # 有两个实根
                x1 = (-b-math.sqrt(delta))/(2*a)
                x2 = (-b+math.sqrt(delta))/(2*a)
                print(f'方程有两个不同的实数根:x1={x1:.2f}, x2={x2:.2f}')
                return x1, x2
            else:
                # print(f'方程无解') # 一下的代码是高中之后的知识。
                # delta < 0有两个虚根
                real_part = -b / (2*a)
                imaginary_part = math.sqrt(-delta) / (2*a)
                x1 = complex(real_part, imaginary_part)
                x2 = complex(real_part, -imaginary_part)
                print(f'方程有两个不同的虚数根:x1={x1:.2f}, x2={x2:.2f}')
    except TypeError as e:
        print(f"发生错误：{e}")


solve_equation(0, -2, 6)
solve_equation(1, -2, 1)
solve_equation(1, -8, 12)
solve_equation(1, -2, 2)

solve_equation("a", -3, 2)
```

#### 自定义异常

```python
class NegativeNumberError(Exception):
    pass

def check_positive(num):
    if num < 0:
        raise NegativeNumberError("不允许负数！")

try:
    check_positive(-5)
except NegativeNumberError as e:
    print(e)
```

##### 代码调试

调试三种错误
- 语法错误，比如少写了一个引号。指开发人员编写了不符合Python语法的代码所引起的错误。
- 运行时错误，比如用一个数去除以零。代码运行期间发生的错误称为异常或运行时错误。

1. 在怀疑出错的地方插入print()语句，输出变量值和数据类型
1. 检查某个 assert 语句 条件是否为真。条件为假，会引发 AssertionError
1. 调试工具 vscode 或PyCharm等
