**本教程以macOS系统为例，讲解python配置环境并且打开juypter.**
# 一、打开MacBook的终端terminal：



输入：`python3 --version`



如果没有显示版本号，用homebrew安装，指令为：
`brew install python3`

# 二、安装vscode必备扩展
打开 VS Code，快捷键 Cmd+Shift+X 进入扩展面板
安装两个官方扩展：
Python（微软官方）
Jupyter（微软官方）
安装完重启 VSCode





# 三、创建项目 + 虚拟环境
新建一个文件夹（比如 PythonLearn），用 VS Code 打开它
打开 VS Code 终端：
执行命令创建虚拟环境：
`python3 -m venv .venv`
激活虚拟环境：
`source.venv/bin/active`
激活成功，终端前会出现（.venv）

在虚拟环境内安装Jupyter依赖
`pip install ipykernel jupyter`

# 四、绑定VSCode解释器
快捷键 Cmd+Shift+P 打开命令面板
输入并选择：Python: Select Interpreter
在列表里选中：
`./ .venv/bin/python3`
选中后，VS Code 右下角会显示对应 Python 版本。

# 五、新建 & 运行 Jupyter Notebook
左上角点文件 → 新建文件，保存为 test.ipynb（后缀必须 .ipynb）
文件打开后，右上角点击 选择内核
选择：Python 3 (ipykernel)
单元格输入代码测试：
运行
```
# 定义函数
def add(a, b):
    return a + b

# 调用函数
print(add(6, 6))
```
按 Shift + Enter 运行，输出 12 即全部配置完成。
