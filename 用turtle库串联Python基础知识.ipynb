# 用turtle库串联Python基础知识

## 什么是turtle

- Turtle 是Python标准库中的一个绘图模块，专门设计用于帮助初学者学习编程。
- 它通过“画乌龟”来进行图形绘制。就像你控制一只乌龟在屏幕上画画，乌龟可以向前、转弯、改变颜色等，适合用来学习编程的基础知识。

## 如何使用turtle库？

- 导入库：在Python脚本中导入Turtle模块。
- 创建画布：设置屏幕和海龟。
- 编写命令：使用Turtle的命令控制海龟移动和绘制。
- 执行和观察：运行代码，观察海龟在屏幕上的绘制过程。

## 基础语法

1. 导入和初始化
    1. 设置画布属性：title(), bgcolor(), speed()

2. 基本绘图命令
    1. 移动控制：forward()、backward() goto(0, 0)
    2. 转向：left(90) ，right(45)
    3. 画笔控制：penup()抬笔、pendown()落笔
3. 颜色与形状：color()设置颜色，fillcolor() 设置填充颜色
4. 常用方法
    1. shape("turtle")改变海龟图标
    2. xcor() # 获取当前的 x 坐标
    3. clear() # 清除屏幕上的所有绘图


```python
# 导入和初始化
import turtle

# 创建画布和海龟
screen = turtle.Screen()
t = turtle.Turtle()

# 设置画布属性
screen.title("我的绘图")
screen.bgcolor("white")

# 设置海龟路径：
# turtle.speed()默认为 3 ，表示中等速度
t.speed(3)
t.color("blue")
t.pensize(2)
t.hideturtle()  #隐藏海龟光标

#基本绘图指令
#移动：
t.forward(100)
t.backward(50)
t.goto(0, 0) #go to direct target

# turn the sight
t.left(90)
t.right(90)

#control the pen:
t.penup()
t.pendown()

# color and fillcolor
t.color("red")
t.fillcolor("orange")
#fill the shape
t.begin_fill()
t.forward(100)
t.left(90)
t.forward(100)
t.left(90)
t.forward(100)
t.left(90)
t.forward(100)
t.end_fill()

# gain the location
current_x = t.xcor()
y = t.ycor()
t.forward(200)
# clear the screen
t.clear()
# write the words
t.write("had cleared the screen",font=("Arial",12,"normal"))

# set the initial place
t.penup()
t.goto(-300, 0)
t.pendown()

t.left(45)

for _ in range(6):
    current_x = t.xcor()
    print(current_x)
    # use if-elif control the color
    # set the color according to the current_x
    if current_x < -200:
        color = 'red'
    elif -200 <= current_x < -100:
        color = 'blue'
    elif -100 <= current_x < 100:
        color = 'orange'
    elif 100 <= current_x < 200:
        color = 'purple'
    else:
        color = 'yellow'

    t.color(color)
    t.forward(100)
    # write the color name in the located place
    t.write(color, font=("Arial",14,"normal"))
screen.exitonclick()
```

### 循环实现-绘制两个

```python
import turtle

square_turtle = turtle.Turtle()
square_turtle.speed(1)
side_length = 100

for i in range(4):
    square_turtle.forward(side_length)
    square_turtle.left(90)

square_turtle.penup() #use pen without line
square_turtle.forward(side_length*1.5)
square_turtle.pendown()

for i in range(4):
    square_turtle.forward(side_length)
    square_turtle.left(90)

square_turtle.hideturtle()
turtle.done()
```

# 函数实现

代码实现

```python
import turtle

square_turtle = turtle.Turtle()
square_turtle.speed(1)
side_length = 100

def draw_square(side_length):
    for i in range(4):
        square_turtle.forward(side_length)
        square_turtle.left(90)

draw_square(side_length)

square_turtle.penup()
square_turtle.forward(side_length*1.5)
square_turtle.pendown()

draw_square(side_length)# the next square

square_turtle.hideturtle()
turtle.done()
```

这里定义了一个函数 draw_square 来绘制正方形

### 画正多边形

```python
import turtle

t = turtle.Turtle()
t.pensize(3)

radius = 50
for i, sides in enumerate(range(3, 9)):
    t.penup()
    t.goto(-300 + i*120, 0)
    t.pendown()
    t.pencolor(["red", "blue", "green", "purple", "orange", "pink"][i-3])
    t.circle(radius, steps=sides)

t.penup()
t.goto(-300, -150)
t.setheading(-90)
t.pendown()
t.speed(0)

# the waves line
wave_radius = 5
wave_angle = 180
for _ in range(30):  # 30 waves
    t.circle(wave_radius, wave_angle)
    t.circle(-wave_radius, wave_angle)
```

### 保持窗口打开

Turtle 的 screen.mainloop() 和 turtle.done()的区别和联系

turtle.done() 简单的场景

screen.mainloop() 复杂交互场景

##### 1.调用方式

##### 2.事件处理

```python
import turtle

screen = turtle.Screen()
t = turtle.Turtle()

# 设置窗口属性
screen.title("鼠标点击")
screen.bgcolor("lightblue")
screen.setup(500, 500)

def handle_click(x, y):
    t.goto(x, y)

screen.onclick(handle_click)  # 绑定鼠标点击事件
screen.mainloop()
```

###### 交互式绘图

```python
import turtle

t = turtle.Turtle()

def move_up():
    t.setheading(90)
    t.forward(10)

def move_down():
    t.setheading(270)
    t.forward(10)

def move_left():
    t.setheading(180)
    t.forward(10)

def move_right():
    t.setheading(0)
    t.forward(10)

screen = turtle.Screen()
screen.listen()
screen.onkey(move_up, "Up")
screen.onkey(move_down, "Down")
screen.onkey(move_left, "Left")
screen.onkey(move_right, "Right")
screen.mainloop()
```
